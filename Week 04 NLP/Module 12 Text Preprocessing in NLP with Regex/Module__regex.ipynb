# Digit Character Class

```python
txt = "that will be 59 dollars"

x = re.findall(r"\d", txt)

print(x)
```

```python
txt = "that will be 59 dollars"

x = re.findall(r"\d+", txt)

print(x)
```

```python
# Match 2, 3

txt = "that will be 190 dollars. My age is 23"

x = re.findall(r"\d{3}", txt)

print(x)
```

```python
# Match 2 or 3

txt = "that will be 190 1 23 dollars. My age is 23"

x = re.findall(r"\d{2,3}", txt) # Starting and ending

print(x)
```

```python
txt = "this is my number 01701-391505"
x = re.findall(r"\d{5}-\d{6}", txt)
x
```

```python
txt = "this is my number +8801701-391505"
x = re.findall(r"\+880\d{4}-\d{6}", txt)

print(x)
```

```python
# At the beginning

txt = "123 123 this is my number 01701-391505"

x = re.findall(r"^\d+", txt)

print(x)
```

```python
# At the end

txt = "123 123 this is my number 01701-391505"

x = re.findall(r"\d+$", txt)

print(x)
```

# Wildcard character .

```python
txt = "hello"

x = re.findall(r".", txt)
print(x)
```

```python
txt = "hellohe00o"

x = re.findall(r"he..o", txt)
print(x)
```

```python
txt = "akibhasan011@gmail.com"

x = re.findall(r".*@.*\..*", txt)

print(x)
```

```python
txt = "akibhasan011@gmail.com batman@gmail.com xyz@gmail.com dog@yahoo.com"

x = re.findall(r".*@.*.com$", txt)

print(x)
```

```python

import re

txt = "akibhasan011@@gmail.com batman@gmail.com xyz@gmail.com dog@yahoo.com"

x = re.findall(r"\b[\w.+-]+@[\w.-]+\.com\b", txt)

print(x)
```

# Capture and group

```python
txt = "ababab"

x = re.findall(r"ab+", txt)

print(x)
```

```python
txt = "ababab"

x = re.findall(r"(ab)+", txt) # (ab)(ab)(ab)

print(x)
```

```python
# Email

txt = "abc@gmail.com"

x = re.findall(r"(\w+)@(\w+)\.com", txt)

print(x)

for first, second in x :
  print(first, second)
```

```python
# Name, age

txt = "john 25, Alice 30, Bob 18"

x = re.findall(r"(\w+)\s(\d+)", txt)

print(x)

for name, age in x :
  print(name, age)
```

# Search

```python
txt = "My lucky number is 25 19 22 1"

x = re.search(r"\d+", txt)

# print(x)

# print(x.start())
# print(x.end())

print(x.group())
```

```python
# Email

txt = "contact us at support@phitron.io , you can also email us admin@yahoo.com"

match = re.search(r"\w+@\w+\.\w+", txt)

print(match.group())
```

# Split

```python
txt = "apple, banana, orange"

x = re.split(r",", txt)

print(x)
```

# Sub

```python
txt = "I have 2 cats and 3 dogs"

x = re.sub(r"\d+", "#", txt)

print(x)
```

# Compile

```python
txt = "I have 2 cats and 15 dogs"
```

```python
pattern = re.compile(r"\d+")
```

```python
# 1: Compile + findall()

x = pattern.findall(txt)

print(x)
```

```python
# 2 compile + search

x = pattern.search(txt)

print(x.group())
```

```python
# Compile + sub

x = pattern.sub('#', txt)

print(x)
```

```python
# Compile + split

pattern = re.compile(r",")

txt = "apple, banana, orange"

x = pattern.split(txt)
print(x)
```
